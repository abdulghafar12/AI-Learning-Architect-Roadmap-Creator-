# **Step 4.1 —  Validate RoadMap**

```python
def validate_roadmap(data):

    required_fields = [
        "domain",
        "level",
        "goal",
        "available_time",
        "deadline",
        "overview",
        "phases",
        "projects",
        "skills_to_master",
        "topics_to_postpone",
        "final_advice"
    ]

    # Check that the main object is a dictionary
    if not isinstance(data, dict):
        return False, "Roadmap is not a dictionary."

    # Check required fields
    for field in required_fields:
        if field not in data:
            return False, f"Missing field: {field}"

    # Check data types
    if not isinstance(data["phases"], list):
        return False, "'phases' must be a list."

    if not isinstance(data["projects"], list):
        return False, "'projects' must be a list."

    if not isinstance(data["skills_to_master"], list):
        return False, "'skills_to_master' must be a list."

    if not isinstance(data["topics_to_postpone"], list):
        return False, "'topics_to_postpone' must be a list."

    # Check phases
    for phase in data["phases"]:

        if not isinstance(phase, dict):
            return False, "Each phase must be a dictionary."

        phase_fields = [
            "phase_number",
            "phase_name",
            "description",
            "estimated_duration",
            "topics"
        ]

        for field in phase_fields:
            if field not in phase:
                return False, f"Phase is missing field: {field}"

        if not isinstance(phase["topics"], list):
            return False, "Phase topics must be a list."

        # Check topics
        for topic in phase["topics"]:

            if not isinstance(topic, dict):
                return False, "Each topic must be a dictionary."

            topic_fields = [
                "topic_name",
                "importance",
                "estimated_time",
                "why_learn",
                "prerequisites",
                "practice"
            ]

            for field in topic_fields:
                if field not in topic:
                    return False, f"Topic is missing field: {field}"

            if not isinstance(topic["prerequisites"], list):
                return False, "Topic prerequisites must be a list."

            if not isinstance(topic["practice"], list):
                return False, "Topic practice must be a list."

    return True, "Roadmap structure is valid."
```

# **Step 5.1 — Build the real generate_roadmap()**

```python
def generate_roadmap(domain, level, goal, available_time, deadline):

    prompt = f"""
Create a personalized learning roadmap.

Student:
- Domain: {domain}
- Level: {level}
- Goal: {goal}
- Available time: {available_time}
- Deadline: {deadline}

Return ONLY valid JSON.

The JSON MUST have exactly these top-level fields:

domain
level
goal
available_time
deadline
overview
phases
projects
skills_to_master
topics_to_postpone
final_advice

Structure:

phases:
[
  {{
    "phase_number": 1,
    "phase_name": "...",
    "description": "...",
    "estimated_duration": "...",
    "topics": [
      {{
        "topic_name": "...",
        "importance": "high",
        "estimated_time": "...",
        "why_learn": "...",
        "prerequisites": [],
        "practice": []
      }}
    ]
  }}
]

projects:
[
  {{
    "project_name": "...",
    "difficulty": "beginner",
    "description": "...",
    "skills_practiced": []
  }}
]

topics_to_postpone:
[
  {{
    "topic": "...",
    "reason": "..."
  }}
]

Rules:
- Create 3 to 5 phases.
- Keep the roadmap concise.
- Make it realistic for the deadline.
- Keep topic descriptions short.
- Include practical exercises.
- Include useful projects.
- importance must be high, medium, or low.
- difficulty must be beginner, intermediate, or advanced.
- Return valid JSON only.
- Do not use markdown.
- Do not add explanations outside JSON.
"""

    response = client.chat.completions.create(

        model="openai/gpt-oss-120b",

        messages=[
            {
                "role": "system",
                "content": """
You are an expert learning curriculum designer.

Return ONLY valid JSON.
Follow the requested structure exactly.
Do not use markdown.
Do not add text outside the JSON object.
"""
            },
            {
                "role": "user",
                "content": prompt
            }
        ],

        response_format={
            "type": "json_object"
        },

        max_completion_tokens=3000,

        reasoning_effort="low"
    )

    content = response.choices[0].message.content

    if not content:
        raise ValueError("AI returned an empty response.")

    roadmap = json.loads(content)

    if not isinstance(roadmap, dict):
        raise ValueError("AI returned an invalid roadmap.")

    # Normalize importance
    for phase in roadmap.get("phases", []):
        for topic in phase.get("topics", []):

            importance = str(
                topic.get("importance", "medium")
            ).lower().strip()

            if importance not in ["high", "medium", "low"]:
                importance = "medium"

            topic["importance"] = importance

    # Normalize difficulty
    for project in roadmap.get("projects", []):

        difficulty = str(
            project.get("difficulty", "beginner")
        ).lower().strip()

        if difficulty not in [
            "beginner",
            "intermediate",
            "advanced"
        ]:
            difficulty = "beginner"

        project["difficulty"] = difficulty

    return roadmap
```

# **Functional Dashboard**

```python
# ============================================================
# AI LEARNING ARCHITECT
# FAST + CLEAN INTERACTIVE DASHBOARD
# Gradio 6.x / Google Colab
#
# IMPORTANT:
# - Uses existing generate_roadmap()
# - Uses existing render_dashboard()
# - No PDF
# - No Download button
# - No roadmap State
# - One Generate callback only
# ============================================================

import gradio as gr
import html
import time


# ============================================================
# FAST ROADMAP GENERATION
# ============================================================

def generate_from_ui(
    domain,
    level,
    goal,
    available_time,
    deadline
):
    """
    Single callback:
        User Input
            ↓
        generate_roadmap()
            ↓
        render_dashboard()
            ↓
        HTML Dashboard
    """

    start_time = time.time()

    try:

        # ----------------------------------------------------
        # VALIDATION
        # ----------------------------------------------------

        domain = str(domain).strip() if domain else ""
        goal = str(goal).strip() if goal else ""
        available_time = str(available_time).strip() if available_time else ""
        deadline = str(deadline).strip() if deadline else ""

        if not domain:
            return """
            <div class="error-box">
                <h3>⚠️ Domain Required</h3>
                <p>Please enter the subject or domain you want to learn.</p>
            </div>
            """

        if not goal:
            return """
            <div class="error-box">
                <h3>⚠️ Learning Goal Required</h3>
                <p>Please describe what you want to achieve.</p>
            </div>
            """

        if not available_time:
            return """
            <div class="error-box">
                <h3>⚠️ Available Time Required</h3>
                <p>Please enter how much time you can study.</p>
            </div>
            """

        if not deadline:
            return """
            <div class="error-box">
                <h3>⚠️ Deadline Required</h3>
                <p>Please enter your target deadline.</p>
            </div>
            """

        # ----------------------------------------------------
        # NORMALIZE INPUT
        # ----------------------------------------------------

        # If user enters only "4", convert it to
        # "4 hours per day"
        if available_time.isdigit():
            available_time = f"{available_time} hours per day"

        # If user enters only "60", convert it to
        # "60 days"
        if deadline.isdigit():
            deadline = f"{deadline} days"

        print("\n" + "=" * 60)
        print("🚀 GENERATING ROADMAP")
        print("=" * 60)
        print(f"Domain: {domain}")
        print(f"Level: {level}")
        print(f"Goal: {goal}")
        print(f"Available Time: {available_time}")
        print(f"Deadline: {deadline}")
        print("-" * 60)

        # ----------------------------------------------------
        # AI GENERATION
        # ----------------------------------------------------

        ai_start = time.time()

        roadmap = generate_roadmap(
            domain=domain,
            level=level,
            goal=goal,
            available_time=available_time,
            deadline=deadline
        )

        ai_time = time.time() - ai_start

        print(f"✅ AI generation: {ai_time:.2f} seconds")

        # ----------------------------------------------------
        # CHECK RESULT
        # ----------------------------------------------------

        if not isinstance(roadmap, dict):
            raise ValueError(
                "The AI returned an invalid roadmap format."
            )

        # ----------------------------------------------------
        # DASHBOARD RENDERING
        # ----------------------------------------------------

        render_start = time.time()

        dashboard_html = render_dashboard(roadmap)

        render_time = time.time() - render_start

        print(f"✅ Dashboard rendering: {render_time:.2f} seconds")

        # ----------------------------------------------------
        # TOTAL TIME
        # ----------------------------------------------------

        total_time = time.time() - start_time

        print(f"⚡ TOTAL TIME: {total_time:.2f} seconds")
        print("=" * 60 + "\n")

        # ----------------------------------------------------
        # RETURN ONLY HTML
        # ----------------------------------------------------

        return dashboard_html

    except Exception as e:

        total_time = time.time() - start_time

        print("\n❌ ROADMAP ERROR")
        print(repr(e))
        print(f"Failed after {total_time:.2f} seconds")
        print("=" * 60 + "\n")

        error_message = html.escape(str(e))

        return f"""
        <div class="error-box">
            <div class="error-icon">⚠️</div>

            <h3>Roadmap Generation Failed</h3>

            <p>
                Something went wrong while creating your roadmap.
            </p>

            <div class="error-details">
                {error_message}
            </div>

            <p class="error-help">
                Please try again. If the problem continues,
                check the notebook output above.
            </p>
        </div>
        """


# ============================================================
# CUSTOM CSS
# ============================================================

custom_css = """

/* =========================================================
   MAIN CONTAINER
   ========================================================= */

#app-container {
    max-width: 1400px !important;
    margin: 0 auto !important;
}


/* =========================================================
   TITLE
   ========================================================= */

.app-title {
    text-align: center;
    padding: 10px 0 20px 0;
}

.app-title h1 {
    font-size: 34px !important;
    font-weight: 800 !important;
    margin-bottom: 6px !important;
}

.app-title p {
    font-size: 16px !important;
    color: #64748B !important;
}


/* =========================================================
   INPUT PANEL
   ========================================================= */

.input-panel {
    border: 1px solid #0F766E !important;
    border-radius: 18px !important;
    padding: 22px !important;
    background: #ffffff !important;
}

.input-panel label {
    font-weight: 600 !important;
}

.input-panel input,
.input-panel textarea,
.input-panel select {
    border: 1px solid #0F766E !important;
}


/* =========================================================
   GENERATE BUTTON
   ========================================================= */

.generate-button {
    background: #0F766E !important;
    color: #ffffff !important;
    border: 1px solid #0F766E !important;
    font-weight: 700 !important;
    border-radius: 12px !important;
    min-height: 50px !important;
    font-size: 16px !important;
}

.generate-button:hover {
    opacity: 0.92 !important;
}


/* =========================================================
   DASHBOARD
   ========================================================= */

.dashboard-container {
    margin-top: 15px !important;
    border-radius: 18px !important;
}


/* =========================================================
   ERROR
   ========================================================= */

.error-box {
    padding: 28px;
    margin: 10px 0;
    border: 1px solid #DC2626;
    border-radius: 16px;
    background: #FFF7F7;
    color: #1F2937;
    font-family: Arial, sans-serif;
}

.error-box h3 {
    color: #B91C1C;
    margin-top: 0;
}

.error-icon {
    font-size: 38px;
    margin-bottom: 8px;
}

.error-details {
    margin-top: 15px;
    padding: 12px;
    border-radius: 8px;
    background: #FEE2E2;
    color: #991B1B;
    font-family: monospace;
    font-size: 13px;
    overflow-wrap: break-word;
}

.error-help {
    margin-top: 15px;
    color: #475569;
}


/* =========================================================
   FOOTER
   ========================================================= */

.footer-text {
    text-align: center;
    color: #64748B !important;
    font-size: 13px !important;
    margin-top: 25px !important;
    padding-bottom: 15px;
}

"""


# ============================================================
# GRADIO INTERFACE
# ============================================================

with gr.Blocks(
    title="AI Learning Architect"
) as app:

    # --------------------------------------------------------
    # MAIN CONTAINER
    # --------------------------------------------------------

    with gr.Column(elem_id="app-container"):

        # ----------------------------------------------------
        # HEADER
        # ----------------------------------------------------

        gr.HTML(
            """
            <div class="app-title">
                <h1>🎓 AI Learning Architect</h1>
                <p>
                    Build a personalized learning roadmap powered by AI.
                </p>
            </div>
            """
        )

        # ----------------------------------------------------
        # INPUT SECTION
        # ----------------------------------------------------

        with gr.Group(elem_classes="input-panel"):

            gr.Markdown(
                "## 🧑‍🎓 Tell us about your learning goal"
            )

            # -----------------------------------------------
            # DOMAIN + LEVEL
            # -----------------------------------------------

            with gr.Row():

                domain_input = gr.Textbox(
                    label="📚 Domain / Subject",
                    placeholder="Example: Machine Learning",
                    scale=2
                )

                level_input = gr.Dropdown(
                    choices=[
                        "Beginner",
                        "Intermediate",
                        "Advanced"
                    ],
                    value="Beginner",
                    label="🎯 Current Level",
                    scale=1
                )

            # -----------------------------------------------
            # GOAL
            # -----------------------------------------------

            goal_input = gr.Textbox(
                label="🚀 Learning Goal",
                placeholder=(
                    "Example: Reach intermediate level "
                    "and build practical projects"
                ),
                lines=3
            )

            # -----------------------------------------------
            # TIME + DEADLINE
            # -----------------------------------------------

            with gr.Row():

                available_time_input = gr.Textbox(
                    label="⏱ Available Time",
                    placeholder="Example: 4 hours per day"
                )

                deadline_input = gr.Textbox(
                    label="📅 Deadline",
                    placeholder="Example: 60 days"
                )

            # -----------------------------------------------
            # GENERATE BUTTON
            # -----------------------------------------------

            generate_button = gr.Button(
                "🚀 Generate My Roadmap",
                variant="primary",
                elem_classes="generate-button"
            )

        # ----------------------------------------------------
        # DASHBOARD TITLE
        # ----------------------------------------------------

        gr.Markdown(
            "## 🗺️ Your Personalized Roadmap"
        )

        # ----------------------------------------------------
        # DASHBOARD OUTPUT
        # ----------------------------------------------------

        dashboard_output = gr.HTML(
            value="""
            <div style="
                padding:40px;
                text-align:center;
                border:1px solid #0F766E;
                border-radius:18px;
                background:#ffffff;
                color:#1F2937;
                font-family:Arial,sans-serif;
            ">

                <div style="
                    font-size:42px;
                    margin-bottom:12px;
                ">
                    🧠
                </div>

                <h3 style="
                    color:#0F766E;
                    margin-bottom:8px;
                ">
                    Your roadmap is waiting
                </h3>

                <p style="
                    color:#475569;
                    margin:0;
                ">
                    Enter your learning details above
                    and click
                    <b>Generate My Roadmap</b>.
                </p>

            </div>
            """,
            elem_classes="dashboard-container"
        )

        # ----------------------------------------------------
        # GENERATE EVENT
        # ----------------------------------------------------
        #
        # IMPORTANT:
        # Only ONE output.
        #
        # We do not save the roadmap in gr.State because
        # there is no PDF/download functionality anymore.
        #
        # show_progress="minimal" keeps the UI lighter than
        # the previous "full" progress handling.
        #

        generate_button.click(
            fn=generate_from_ui,
            inputs=[
                domain_input,
                level_input,
                goal_input,
                available_time_input,
                deadline_input
            ],
            outputs=dashboard_output,
            show_progress="minimal"
        )

        # ----------------------------------------------------
        # FOOTER
        # ----------------------------------------------------

        gr.HTML(
            """
            <div class="footer-text">
                AI Learning Architect
                • Personalized learning, structured by AI.
            </div>
            """
        )


# ============================================================
# LAUNCH
# ============================================================

app.launch(
    share=True,
    debug=False,
    css=custom_css
)
```
